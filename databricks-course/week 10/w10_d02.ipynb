```sql
CREATE SCHEMA IF NOT EXISTS workspace.coffee_reports
COMMENT 'Read-only views for the analysts';
```

```sql
CREATE OR REPLACE VIEW workspace.coffee_reports.revenue_by_channel AS
SELECT channel, lines, orders, revenue_eur
FROM workspace.coffee_pipeline.gold_revenue_by_channel;
```

```sql
CREATE OR REPLACE VIEW workspace.coffee_reports.product_revenue AS
SELECT product, category, qty, revenue_eur
FROM workspace.coffee_pipeline.gold_product_revenue;
```

```sql
SELECT * FROM workspace.coffee_reports.revenue_by_channel ORDER BY channel;
```

```sql
CREATE OR REPLACE VIEW workspace.coffee_reports.revenue_total AS
SELECT ROUND(SUM(revenue_eur), 2) AS revenue_eur
FROM workspace.coffee_pipeline.gold_revenue_by_channel;
```
