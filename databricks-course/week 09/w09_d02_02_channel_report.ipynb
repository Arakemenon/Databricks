```python
from pyspark.sql.functions import col, count, sum as spark_sum, round as spark_round

dbutils.widgets.text("channel", "shop")
channel = dbutils.widgets.get("channel")

report = (spark.table("workspace.coffee.silver_order_lines")
    .filter(col("channel") == channel)
    .agg(count("*").alias("lines"),
         spark_round(spark_sum("line_amount_eur"), 2).alias("revenue_eur"))
    .first())

print(f"{channel}: {report['lines']} lines, {report['revenue_eur']} EUR")
```
